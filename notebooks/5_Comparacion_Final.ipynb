## 2. Carga de datos críticos y vectorización TF-IDF

Base común para la **PARTE A** (modelos en datos críticos).
Dataset: `rad_criticos.xlsx` · Split 80/20 · `random_state=42`

```python
# =============================================================================
# CARGA DE DATOS
# =============================================================================

df = pd.read_excel('../data/rad_criticos.xlsx')

print(f"Dataset cargado: {df.shape[0]} muestras, {df.shape[1]} columnas")
```

```python
# =============================================================================
# ETIQUETADO DE PATOLOGÍAS (generado automáticamente desde Hallazgos + Opinión)
# =============================================================================

def detectar_patologia_robusta(texto, keywords_positivos, keywords_negativos=None):
    """Detecta patologías con análisis de contexto robusto (ventana de negación)."""
    if not isinstance(texto, str):
        return 0
    texto = texto.lower()
    keywords_negativos = keywords_negativos or [
        'antiguo','antigua','cronico','cronica','crónico','crónica',
        'viejo','vieja','residual','residuales','secuela','secuelar','secuelas',
        'previo','previa','previos','previas','conocido','conocida',
        'sin cambios','estable','estables','resolucion','resolución'
    ]
    negaciones_contextuales = [
        'sin','descarta','descartado','descartada','negativo','negativa',
        'ausencia','ausente','ausentes','sin evidencia','sin signos',
        'no se observa','no se observan','no se evidencia','no se evidencian',
        'no se identifica','no se identifican','libre de',
        'negativo para','negativa para'
    ]
    keywords_ordenadas = sorted(keywords_positivos, key=len, reverse=True)
    oraciones = re.split(r'[.;]', texto)
    for oracion in oraciones:
        oracion = oracion.strip()
        keyword_encontrada = next((kw for kw in keywords_ordenadas if kw in oracion), None)
        if not keyword_encontrada:
            continue
        if re.search(re.escape(keyword_encontrada) + r'\s*:\s*no\b', oracion):
            continue
        if re.search(re.escape(keyword_encontrada) + r'\s*:[^.;]*?\b(no|negativo|negativa|ausente)\b', oracion):
            continue
        idx_kw = oracion.find(keyword_encontrada)
        segmento_previo = oracion[max(0, idx_kw-50):idx_kw]
        if any(neg in segmento_previo for neg in negaciones_contextuales):
            continue
        if re.search(r'sin\s+(hallazgos?|evidencia|signos?)\s+(indirectos?\s+)?(de\s+)?' + re.escape(keyword_encontrada), oracion):
            continue
        if any(ant in oracion for ant in keywords_negativos):
            continue
        if 'ya conocido' in oracion or 'ya conocida' in oracion:
            continue
        if 'sin cambios' in oracion:
            continue
        if 'resolucion' in oracion or 'resolución' in oracion:
            continue
        return 1
    return 0

texto_etiquetado = df['Hallazgos'] + ' ' + df['Opinión']

keywords = {
    'acv': [
        'infarto cerebral','infarto isquemico','accidente cerebrovascular',
        'hipodensidad sugestiva de isquemia','zona de isquemia',
        'area de isquemia','evento isquemico','ictus','acv','isquemia'
    ],
    'hemorragia': [
        'hemorragia subaracnoidea','hemorragia intraparenquimatosa',
        'hemorragia intraventricular','hematoma subdural','hematoma epidural',
        'hematoma intraparenquimatoso','sangrado intraparenquimatoso',
        'coleccion hematica','hemorragia','hematoma','sangrado'
    ],
    'desviacion_linea_media': [
        'desviacion de la linea media','desviacion linea media','linea media desplazada',
        'desplazamiento de la linea media','efecto de masa con desviacion',
        'shift de linea media','herniacion subfalcina','hernia subfalcina'
    ],
    'fractura_compleja_craneo': [
        'fractura con hundimiento','fractura hundimiento','fractura deprimida',
        'fractura conminuta','fractura compleja','fractura de base de craneo',
        'fractura craneofacial','fractura multiple de craneo'
    ]
}

for pat, kws in keywords.items():
    df[pat] = texto_etiquetado.apply(lambda t: detectar_patologia_robusta(t, kws))

print("Distribución de etiquetas generadas:")
for p in ['acv', 'hemorragia', 'desviacion_linea_media', 'fractura_compleja_craneo']:
    n = df[p].sum()
    print(f"  {p:12s}: {n:5d}  ({100*n/len(df):.1f}%)")
```

```python
# =============================================================================
# DISTRIBUCIÓN + SPLIT + VECTORIZACIÓN TF-IDF
# =============================================================================

print("\nDistribución de clases generadas (positivos):")
for pat in PATOLOGIAS:
    n_pos = df[pat].sum()
    print(f"  {pat:12s}: {n_pos:4d} positivos ({n_pos/len(df)*100:.1f}%)")

X_text = df['Hallazgos'].astype(str)
y = df[PATOLOGIAS].values

# Split estratificado 80/20 — idéntico al de los cuadernos originales
X_text_train, X_text_test, y_train, y_test = train_test_split(
    X_text, y,
    test_size=0.2,
    random_state=42,
    stratify=df[PATOLOGIAS[0]].values
)

print(f"\nTrain: {len(X_text_train)} muestras | Test: {len(X_text_test)} muestras")

# Vectorización TF-IDF — misma configuración que cuadernos 2 y 5
stopwords_es = ['de','la','y','el','en','del','con','se','las','los','por','para',
                'al','lo','un','una','es','su','sus','que','a','o','sin','no',
                'izquierdo','izquierda','derecha','derecho','mm']

vectorizer = TfidfVectorizer(
    max_features=200,
    min_df=5,
    max_df=0.8,
    ngram_range=(1, 2),
    stop_words=stopwords_es
)

X_train = vectorizer.fit_transform(X_text_train)
X_test  = vectorizer.transform(X_text_test)

print(f"TF-IDF: {X_train.shape[1]} features")
```

### A1. SVM Linear (Críticos)
Configuración: `class_weight='balanced'`, `max_iter=1000`, `random_state=42`.

```python
# =============================================================================
# SVM LINEAR — configuración idéntica al cuaderno 1
# Entrenado en datos críticos (rad_criticos.xlsx, split 80/20)
# =============================================================================

print("=" * 70)
print("ENTRENAMIENTO: SVM LINEAR (Críticos)")
print("=" * 70)

svm_base = LinearSVC(class_weight='balanced', max_iter=1000, random_state=42)

resultados_svm = []
predicciones_svm = {}

for idx, pat in enumerate(PATOLOGIAS):
    y_train_pat = y_train[:, idx]
    y_test_pat  = y_test[:, idx]

    modelo = clone(svm_base)
    modelo.fit(X_train, y_train_pat)
    y_pred = modelo.predict(X_test)

    precision = precision_score(y_test_pat, y_pred, zero_division=0)
    recall    = recall_score(y_test_pat, y_pred, zero_division=0)
    f1        = f1_score(y_test_pat, y_pred, zero_division=0)
    kappa     = cohen_kappa_score(y_test_pat, y_pred)

    predicciones_svm[pat] = {'y_test': y_test_pat, 'y_pred': y_pred}
    resultados_svm.append({
        'Patología': pat,
        'Precision': precision,
        'Recall': recall,
        'F1': f1,
        'Kappa': kappa
    })
    print(f"  {pat:30s} → F1: {f1:.3f} | Recall: {recall:.3f} | Precision: {precision:.3f} | Kappa: {kappa:.3f}")

f1_macro_svm        = np.mean([r['F1']       for r in resultados_svm])
recall_macro_svm    = np.mean([r['Recall']   for r in resultados_svm])
precision_macro_svm = np.mean([r['Precision']for r in resultados_svm])
kappa_macro_svm     = np.mean([r['Kappa']    for r in resultados_svm])

print(f"\n{'─'*60}")
print(f"  {'MACRO':30s} → F1: {f1_macro_svm:.3f} | Recall: {recall_macro_svm:.3f} | "
      f"Precision: {precision_macro_svm:.3f} | Kappa: {kappa_macro_svm:.3f}")
```

### A2. XGBoost (Críticos)
Configuración: `scale_pos_weight` calculado por patología desde el conjunto de entrenamiento, `n_estimators=100`, `max_depth=6`.

```python
# =============================================================================
# XGBOOST CON scale_pos_weight — configuración idéntica al cuaderno 4
# =============================================================================

print("=" * 70)
print("ENTRENAMIENTO: XGBoost (scale_pos_weight)")
print("=" * 70)

resultados_xgb = []
predicciones_xgb = {}

for idx, pat in enumerate(PATOLOGIAS):
    y_train_pat = y_train[:, idx]
    y_test_pat  = y_test[:, idx]
    
    # scale_pos_weight calculado solo desde train
    n_pos = y_train_pat.sum()
    n_neg = len(y_train_pat) - n_pos
    spw   = n_neg / n_pos
    
    modelo = XGBClassifier(
        n_estimators=100,
        max_depth=6,
        learning_rate=0.3,
        scale_pos_weight=spw,
        eval_metric='logloss',
        use_label_encoder=False,
        random_state=42,
        n_jobs=-1
    )
    modelo.fit(X_train, y_train_pat)
    y_pred = modelo.predict(X_test)
    
    precision = precision_score(y_test_pat, y_pred, zero_division=0)
    recall    = recall_score(y_test_pat, y_pred, zero_division=0)
    f1        = f1_score(y_test_pat, y_pred, zero_division=0)
    kappa     = cohen_kappa_score(y_test_pat, y_pred)
    
    predicciones_xgb[pat] = {'y_test': y_test_pat, 'y_pred': y_pred}
    resultados_xgb.append({
        'Patología': pat,
        'Precision': precision,
        'Recall': recall,
        'F1': f1,
        'Kappa': kappa
    })
    print(f"  {pat:12s} → F1: {f1:.3f} | Recall: {recall:.3f} | Precision: {precision:.3f} | Kappa: {kappa:.3f} [SPW={spw:.2f}]")

f1_macro_xgb        = np.mean([r['F1'] for r in resultados_xgb])
recall_macro_xgb    = np.mean([r['Recall'] for r in resultados_xgb])
precision_macro_xgb = np.mean([r['Precision'] for r in resultados_xgb])
kappa_macro_xgb     = np.mean([r['Kappa'] for r in resultados_xgb])

print(f"\n{'─'*50}")
print(f"  {'MACRO':12s} → F1: {f1_macro_xgb:.3f} | Recall: {recall_macro_xgb:.3f} | "
      f"Precision: {precision_macro_xgb:.3f} | Kappa: {kappa_macro_xgb:.3f}")
```

## 7. Predicción final, métricas multilabel y análisis de errores

Funciones listas para usar con `models`, `vectorizer`, `y_true/y_pred` y los DataFrames ya cargados.

```python
from concurrent.futures import ThreadPoolExecutor
from sklearn.metrics import accuracy_score, hamming_loss, f1_score
import numpy as np
import pandas as pd
import re

LABELS_DEFAULT = ['acv', 'hemorragia', 'desviacion_linea_media', 'fractura_compleja_craneo']
LABEL_DISPLAY = {
    'acv': 'ACV',
    'hemorragia': 'Hemorragia',
    'desviacion_linea_media': 'Desv. L. Media',
    'fractura_compleja_craneo': 'Fractura Compleja',
}


# =============================================================================
# TAREA 1 — Prediccion final explicita
# =============================================================================

def _sigmoid(x):
    return 1 / (1 + np.exp(-x))


def _predict_one_label(label, model, X_vec, threshold):
    if hasattr(model, 'predict_proba'):
        prob = float(model.predict_proba(X_vec)[0][1])
    elif hasattr(model, 'decision_function'):
        score = model.decision_function(X_vec)
        prob = float(_sigmoid(score)[0])
    else:
        pred = int(model.predict(X_vec)[0])
        prob = float(pred)
    decision = prob >= threshold
    return label, prob, decision


def predict_multilabel(texto, models=None, vectorizer=None,
                       labels=LABELS_DEFAULT, thresholds=None, use_parallel=True):
    if texto is None:
        texto = ''
    texto = str(texto)
    thresholds = thresholds or {}

    # resolve defaults from globals if not provided
    if models is None:
        if 'models' in globals():
            models = globals()['models']
        else:
            raise NameError("models is not defined. Pass 'models' to predict_multilabel or define it in globals.")
    if vectorizer is None:
        if 'vectorizer' in globals():
            vectorizer = globals()['vectorizer']
        else:
            raise NameError("vectorizer is not defined. Pass 'vectorizer' to predict_multilabel or define it in globals.")

    X_vec = vectorizer.transform([texto])

    def runner(label):
        # support models as dict or as sequence aligned with labels
        if isinstance(models, dict):
            model = models[label]
        elif isinstance(models, (list, tuple)):
            idx = labels.index(label)
            model = models[idx]
        else:
            raise TypeError("models must be a dict or a list/tuple of models")
        threshold = thresholds.get(label, 0.5)
        return _predict_one_label(label, model, X_vec, threshold)

    if use_parallel:
        with ThreadPoolExecutor(max_workers=min(4, max(1, len(labels)))) as ex:
            results = list(ex.map(runner, labels))
    else:
        results = [runner(label) for label in labels]

    salida = {}
    for label, prob, decision in results:
        salida[label] = {'label': label, 'prob': prob, 'decision': bool(decision)}

    for label, prob, decision in results:
        nombre = LABEL_DISPLAY.get(label, label)
        marca = '✓' if decision else ''
        print(f"{nombre:<18} → {'SÍ' if decision else 'NO':<2} (prob: {prob:.2f}) {marca}")

    return salida


# =============================================================================
# TAREA 2 — Metricas multilabel nativas
# =============================================================================

def _interpretar_metrica(nombre, valor):
    if nombre == 'Hamming Loss':
        if valor <= 0.05:
            return 'Excelente (muy baja)'
        if valor <= 0.10:
            return 'Buena (baja)'
        if valor <= 0.20:
            return 'Aceptable'
        return 'Alta (riesgo clinico)'

    if valor >= 0.90:
        return 'Excelente'
    if valor >= 0.80:
        return 'Buena'
    if valor >= 0.70:
        return 'Aceptable'
    return 'Baja (riesgo clinico)'


def multilabel_metrics_df(y_true, y_pred, label_names=None):
    label_names = label_names or LABELS_DEFAULT

    subset_acc = accuracy_score(y_true, y_pred)
    hamming = hamming_loss(y_true, y_pred)
    f1_macro = f1_score(y_true, y_pred, average='macro')
    f1_micro = f1_score(y_true, y_pred, average='micro')
    f1_labels = f1_score(y_true, y_pred, average=None)

    filas = [
        {'Metrica': 'Subset Accuracy', 'Valor': subset_acc,
         'Interpretacion': _interpretar_metrica('Subset Accuracy', subset_acc)},
        {'Metrica': 'Hamming Loss', 'Valor': hamming,
         'Interpretacion': _interpretar_metrica('Hamming Loss', hamming)},
        {'Metrica': 'F1 Macro', 'Valor': f1_macro,
         'Interpretacion': _interpretar_metrica('F1 Macro', f1_macro)},
        {'Metrica': 'F1 Micro', 'Valor': f1_micro,
         'Interpretacion': _interpretar_metrica('F1 Micro', f1_micro)},
    ]

    for nombre, valor in zip(label_names, f1_labels):
        filas.append({
            'Metrica': f"F1-{nombre}",
            'Valor': float(valor),
            'Interpretacion': _interpretar_metrica('F1', float(valor))
        })

    df_metricas = pd.DataFrame(filas)
    print(df_metricas.to_string(index=False))
    return df_metricas


# =============================================================================
# TAREA 3 — Analisis cualitativo de FP y FN
# =============================================================================

NEGATION_PATTERNS = [
    r"\bsin\b", r"\bno\b", r"\bdescarta\b", r"\bdescartado\b",
    r"\bdescartada\b", r"\bnegativo para\b"
]
AMBIGUOUS_PATTERNS = [
    r"\bposible\b", r"\bprobable\b", r"\bsugestivo\b", r"\bsugestiva\b",
    r"no se descarta"
]


def _get_keywords_for_label(label):
    if 'KEYWORDS' in globals() and label in KEYWORDS:
        return KEYWORDS[label]
    return [label.replace('_', ' ')]


def _tiene_negacion(texto, label):
    keywords = _get_keywords_for_label(label)
    if not keywords:
        return False
    kw_pat = '|'.join(map(re.escape, keywords))
    return re.search(rf"({'|'.join(NEGATION_PATTERNS)})\s+({kw_pat})", texto) is not None


def _tiene_ambiguedad(texto):
    return re.search(rf"({'|'.join(AMBIGUOUS_PATTERNS)})", texto) is not None


def _infer_train_counts(label_names):
    if 'y_train' in globals():
        return {label: int(np.sum(y_train[:, idx])) for idx, label in enumerate(label_names)}
    return {label: 0 for label in label_names}


def clasificar_causa(texto, label, train_counts, min_train_cases=50):
    texto = texto.lower()
    if _tiene_negacion(texto, label):
        return 'negacion_no_capturada'
    if _tiene_ambiguedad(texto):
        return 'termino_ambiguo'
    if train_counts.get(label, 0) < min_train_cases:
        return 'soporte_insuficiente'
    return 'otro'


def analizar_fp_fn(textos, y_true, y_pred, label_names=None, train_counts=None, max_examples=3):
    label_names = label_names or LABELS_DEFAULT
    train_counts = train_counts or _infer_train_counts(label_names)

    textos_serie = pd.Series(textos).fillna('').astype(str)
    textos_serie = textos_serie.str.replace(r"\s+", " ", regex=True)

    for idx, label in enumerate(label_names):
        fp_mask = (y_true[:, idx] == 0) & (y_pred[:, idx] == 1)
        fn_mask = (y_true[:, idx] == 1) & (y_pred[:, idx] == 0)

        df_fp = pd.DataFrame({
            'tipo': 'FP',
            'texto': textos_serie[fp_mask].str.slice(0, 300),
        })
        df_fn = pd.DataFrame({
            'tipo': 'FN',
            'texto': textos_serie[fn_mask].str.slice(0, 300),
        })

        df_fp['posible_causa'] = df_fp['texto'].apply(
            lambda t: clasificar_causa(t, label, train_counts)
        )
        df_fn['posible_causa'] = df_fn['texto'].apply(
            lambda t: clasificar_causa(t, label, train_counts)
        )

        print(f"\n=== {LABEL_DISPLAY.get(label, label)} ===")
        print(f"FP: {len(df_fp)} | FN: {len(df_fn)}")

        if len(df_fp) > 0:
            print('\nFP (primeros ejemplos):')
            display(df_fp.head(max_examples))
        else:
            print('\nSin FP en este conjunto.')

        if len(df_fn) > 0:
            print('\nFN (primeros ejemplos):')
            display(df_fn.head(max_examples))
        else:
            print('\nSin FN en este conjunto.')


# =============================================================================
# TAREA 4 — Diagrama de flujo de datos (matplotlib)
# =============================================================================

def _crit_pct_from_df(df_local):
    if df_local is None or 'Hallazgo Crítico' not in df_local.columns:
        return None
    col = df_local['Hallazgo Crítico']
    if col.dtype == bool:
        crit = col.mean()
    elif np.issubdtype(col.dtype, np.number):
        crit = (col > 0).mean()
    else:
        crit = col.astype(str).str.lower().str.contains('si|sí|true|1').mean()
    return float(crit * 100)


def _crit_pct_from_y(y_local):
    if y_local is None:
        return None
    crit = (np.sum(y_local, axis=1) > 0).mean()
    return float(crit * 100)


def _format_node_text(nombre, total, crit_pct):
    total_txt = str(total) if total is not None else 'N/D'
    if crit_pct is None:
        pct_txt = 'Crít: N/D | No crít: N/D'
    else:
        pct_txt = f"Crít: {crit_pct:.1f}% | No crít: {100 - crit_pct:.1f}%"
    return f"{nombre}\n(n={total_txt})\n{pct_txt}"


def plot_flujo_datos(df_original=None, df_rad_criticos=None, y_train=None, y_test=None,
                     total_original=3984, total_rad=952, total_train=761, total_test=190,
                     total_llm=1334, total_ros=None):
    total_original = len(df_original) if df_original is not None else total_original
    total_rad = len(df_rad_criticos) if df_rad_criticos is not None else total_rad
    total_train = len(y_train) if y_train is not None else total_train
    total_test = len(y_test) if y_test is not None else total_test

    pct_original = _crit_pct_from_df(df_original)
    pct_rad = _crit_pct_from_df(df_rad_criticos)
    pct_train = _crit_pct_from_y(y_train)
    pct_test = _crit_pct_from_y(y_test)

    if pct_rad is None:
        pct_rad = 100.0
    if pct_train is None:
        pct_train = 100.0
    if pct_test is None:
        pct_test = 100.0

    pct_llm = 100.0
    pct_ros = 100.0

    COLOR_NOCRIT = '#4EA8DE'
    COLOR_CRIT = '#F27B7B'
    COLOR_PROC = '#6BCB77'

    fig, ax = plt.subplots(figsize=(14, 5))
    ax.set_xlim(0, 1.35)
    ax.set_ylim(0.2, 0.9)
    ax.axis('off')

    def draw_node(x, y, text, color):
        ax.text(
            x, y, text, ha='center', va='center', fontsize=8.5,
            bbox=dict(boxstyle='round,pad=0.4', fc=color, ec='black', lw=1.2)
        )

    def draw_arrow(x1, y1, x2, y2):
        ax.annotate('', xy=(x2, y2), xytext=(x1, y1),
                    arrowprops=dict(arrowstyle='->', lw=1.6, color='#555'))

    # Nodos de datos
    draw_node(0.05, 0.6, _format_node_text('data.xlsx', total_original, pct_original), COLOR_NOCRIT)
    draw_node(0.39, 0.6, _format_node_text('rad_criticos.xlsx', total_rad, pct_rad), COLOR_CRIT)
    draw_node(0.73, 0.72, _format_node_text('train (80%)', total_train, pct_train), COLOR_CRIT)
    draw_node(0.73, 0.40, _format_node_text('test (20%)', total_test, pct_test), COLOR_CRIT)
    draw_node(0.90, 0.72,
              _format_node_text('LLM augmentation', total_llm, pct_llm), COLOR_CRIT)
    draw_node(1.07, 0.72,
              _format_node_text('ROS balanceado', total_ros, pct_ros), COLOR_CRIT)

    # Nodos de proceso
    draw_node(0.22, 0.6, 'Filtro\nHallazgo critico', COLOR_PROC)
    draw_node(0.56, 0.6, 'Split\n80/20', COLOR_PROC)
    draw_node(1.24, 0.72, 'Modelo\nXGBoost', COLOR_PROC)

    # Flechas
    draw_arrow(0.12, 0.6, 0.18, 0.6)
    draw_arrow(0.26, 0.6, 0.35, 0.6)
    draw_arrow(0.43, 0.6, 0.52, 0.6)
    draw_arrow(0.60, 0.62, 0.68, 0.70)
    draw_arrow(0.60, 0.58, 0.68, 0.42)
    draw_arrow(0.78, 0.72, 0.85, 0.72)
    draw_arrow(0.95, 0.72, 1.02, 0.72)
    draw_arrow(1.12, 0.72, 1.20, 0.72)

    ax.set_title('Flujo de datos — Clasificacion TC de craneo', fontsize=12, fontweight='bold')
    plt.tight_layout()
    plt.show()


# Ejemplos de uso (ajusta segun tus variables)
# predict_multilabel("Paciente con hemorragia subaracnoidea...")
# df_metricas = multilabel_metrics_df(y_test, y_pred, label_names=LABELS_DEFAULT)
# analizar_fp_fn(textos, y_test, y_pred, label_names=LABELS_DEFAULT)
# plot_flujo_datos(df_original, df_rad_criticos, y_train, y_test)
```
